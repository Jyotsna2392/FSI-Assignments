```python
#PROBLEM - 1
students = { "Rahul": 91, "Ayesha": 84, "Kiran": 67, "Ali": 45 }
countp=0
countf=0
for name,marks in students.items():
    print('name : ',name)
    print('marks : ',marks)
    if marks >=40:
        print('result : pass')
        countp+=1
    else:
        print('result : Fail')
        countf+=1
print('pass count : ', countp)
print('Fail count : ', countf)
```

```python
#problem - 6
expenses = {} 
while True: 
   name = input("Enter Expense Name (or stop): ") 
   if name.lower() == "stop": 
      break 
   amount = float(input("Enter Amount: ")) 
   expenses[name] = amount 
print("\nExpense Details") 
for item, price in expenses.items(): 
    print(item, ":", price) 
print("Total Expenses:", len(expenses))
```

```python
#problem 11
scores = { 
"Rahul":120, 
"Ali":250, 
"John":180 
}
highest_player=''
highest_score=0
for player,score in scores.items():
    if score > highest_score:
        highest_score=score
        highest_player=player
print('highest player : ',highest_player)
print('highest_score : ', highest_score)
```

```python
#problem - 13 
seats = { 
    "A1": "Available", 
    "A2": "Booked", 
    "A3": "Available", 
    "A4": "Available" 
} 
 
seat = input("Enter Seat Number: ").upper() 
 
if seat in seats: 
 
    if seats[seat] == "Available": 
 
        seats[seat] = "Booked" 
 
        print("Seat Booked Successfully") 
    else: 
        print("Seat Already Booked") 
else: 
   print("Invalid Seat Number") 
print(seats)
```

```python
#problem - 15
items = { 
    "Milk": 50, 
    "Bread": 40, 
    "Eggs": 70, 
    "Rice": 500 
} 
 
bill = 0 
 
while True: 
 
    item = input("Enter Item Name (done to stop): ").title() 
 
    if item.lower() == "done": 
        break 
 
    if item in items: 
 
        bill += items[item] 
        print(item, "Added") 
    else: 
        print("Item Not Available") 
print("Final Bill:", bill)
```

```python
#problem - 20
systems = { 
"Engine": "OK", 
"Fuel": "OK", 
"Navigation": "FAIL", 
"Communication": "OK" 
} 
file = open("mission_log.txt", "w") 
 
for system, status in systems.items(): 
 
    if status == "FAIL": 
 
        print(system, "Failed") 
 
        file.write(system + " Failed\n") 
 
        print("Mission Aborted") 
 
        break 
 
    else: 
 
       file.write("Launch Successful") 
 
file.close() 
```
